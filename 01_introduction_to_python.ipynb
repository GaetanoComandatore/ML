## 6. Lists

Lists are tha main python objects to encapsulate a sequence. Lists can be ordered, they are iterable (i.e. you can loop through lists), they are mutable (i.e. you can change their content) and, most importantly, they can contain multiple data types!

```python
# create an empty list (two ways)
empty_list = []
empty_list = list()
```

```python
# create a list
simpsons = ['homer', 'marge', 'bart']
```

**Examine a list:**

```python
# print element 0
simpsons[0]
```

```python
len(simpsons)
```

**Modify a list (does not return the list):**

```python
# append element to end
simpsons.append('lisa')
simpsons
```

```python
# append multiple elements to end
simpsons.extend(['itchy', 'scratchy'])
simpsons
```

```python
# insert element at index 0 (shifts everything right)
simpsons.insert(0, 'maggie')
simpsons
```

```python
# search for first instance and remove it
simpsons.remove('bart')
simpsons
```

```python
# remove element 0 and return it
simpsons.pop(0)
```

```python
# remove element 0 (does not return it)
del simpsons[0]
simpsons
```

```python
# replace element 0
simpsons[0] = 'krusty'
simpsons
```

```python
# trying to remove a non-existing element will raise an error
try:
    simpsons.remove('bart') # we have already removed the element 'bart'
except ValueError as err:
    print("You can't remove element 'bart' because", err)
```

```python
# concatenate lists (slower than 'extend' method)
neighbors = simpsons + ['ned', 'rod', 'todd']
neighbors
```

```python
# list concatenation returns an element, whilst the 'extend' method does not
print("List returned by '+' (concat. operator) =", neighbors + ['another guy'])
print("List returned by the 'extend' method =", neighbors.extend(['another girl']))
```

```python
# however, the 'extend' method acts 'in-place'.
# Now 'neighbors' will have 'another girl' (added with 'extend')
# but not 'another guy' (added with concatenation)
print("Current neighbors list = ", neighbors)
```

**Define a function with two 'positional arguments' (no default values) and one 'keyword argument' (has a default value):**


```python
def calc(a, b, op='add'):
    if op == 'add':
        return a + b
    elif op == 'sub':
        return a - b
    else:
        print('valid operations are add and sub')
```

```python
# call the function
calc(10, 4, op='add')
```

```python
# unnamed arguments are inferred by position
calc(10, 4, 'add')
```

```python
calc(10, 4, 'sub')
```

```python
calc(10, 4, 'div')
```

```python
# note that you can omit keyword arguments when calling 'calc', since they have
# a default value (in this case, the default value for op is 'add' and will be used)
calc(10, 4)
```

```python
# but you cannot omit positional arguments! As no default value is specified,
# Python needs a proper value to populate a and b! The following will all result
# in errors.
try:
    calc()
except Exception as err:
    print("You've called calc(), but", err)
```

```python
try:
    calc(10, op="add")
except Exception as err:
    print("You've called calc(10, op='add'), but", err)
```

```python
try:
    calc(b=10, op="add")
except Exception as err:
    print("You've called calc(b=10, op='add'), but", err)
```

## 13. For Loops and While Loops

**`range` returns a sequence (Python 3):**

```python
# includes the start value but excludes the stop value
range(0, 3)
```

```python
# default start value is 0
range(3)
```

```python
# third argument is the step value
range(0, 5, 2)
```

**`for` loops:**

```python
# not the recommended style
fruits = ['apple', 'banana', 'cherry']
for i in range(len(fruits)):
    print(fruits[i].upper())
```

```python
# recommended style
for fruit in fruits:
    print(fruit.upper())
```

```python
# iterate through two things at once (using tuple unpacking)
family = {'dad':'homer', 'mom':'marge', 'size':6}
for key, value in family.items():
    print(key, value)
```

```python
# use enumerate if you need to access the index value within the loop
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
# to clarify, the former enumerate loop is equivalent to
for index in range(len(fruits)):
    print(index, fruits[index])
```

```python
# to loop through more than one iterable at a time, use 'zip'
names = ["John", "Mary", "Joe", "Camille"]
surnames = ["Snow", "McDonald's", "Simpson", "Richards"]
for name, surname in zip(names, surnames):
    print(name, surname)
```

**`for`/`else` loop:**

```python
for fruit in fruits:
    if fruit == 'banana':
        print('Found the banana!')
        break    # exit the loop and skip the 'else' block
else:
    # this block executes ONLY if the for loop completes without hitting 'break'
    print("Can't find the banana")
```

```python
# skip specific iterations in loops (this will print all except 'banana')
for fruit in fruits:
  if fruit == 'banana':
    continue # skip the rest of the execution for this iteration, but do not exit the loop
  print(fruit)
```

**`while` loop:**

```python
count = 0
while count < 5:
    print('This will print 5 times')
    count += 1    # equivalent to 'count = count + 1', there is no '++' in python
print(f"Count is {count}, not in the while loop anymore.")
```

## 14. Comprehensions

**List comprehension:**

```python
# for loop to create a list of cubes
nums = [1, 2, 3, 4, 5]
cubes = []
for num in nums:
    cubes.append(num**3)
cubes
```

```python
# equivalent list comprehension
cubes = [num**3 for num in nums]
cubes
```

```python
# for loop to create a list of cubes of even numbers
cubes_of_even = []
for num in nums:
    if num % 2 == 0:
        cubes_of_even.append(num**3)
cubes_of_even
```

```python
# equivalent list comprehension
# syntax: [expression for variable in iterable if condition]
cubes_of_even = [num**3 for num in nums if num % 2 == 0]
cubes_of_even
```

```python
# for loop to cube even numbers and square odd numbers
cubes_and_squares = []
for num in nums:
    if num % 2 == 0:
        cubes_and_squares.append(num**3)
    else:
        cubes_and_squares.append(num**2)
cubes_and_squares
```

```python
# equivalent list comprehension (using a ternary expression)
# syntax: [true_condition if condition else false_condition for variable in iterable]
cubes_and_squares = [num**3 if num % 2 == 0 else num**2 for num in nums]
cubes_and_squares
```

```python
# for loop to flatten a 2d-matrix
matrix = [[1, 2], [3, 4]]
items = []
for row in matrix:
    for item in row:
        items.append(item)
items
```

```python
# equivalent list comprehension
items = [item for row in matrix
              for item in row]
items
```

**Set comprehension:**

```python
fruits = ['apple', 'banana', 'cherry']
unique_lengths = {len(fruit) for fruit in fruits}
unique_lengths
```

**Dictionary comprehension:**

```python
fruit_lengths = {fruit:len(fruit) for fruit in fruits}
fruit_lengths
```

```python
# you can also pair a list/set/dictionary comprehension with built-in methods
# such as enumerate:
fruit_indices = {fruit:index for index, fruit in enumerate(fruits)}
fruit_indices
```

## 15. Map and Filter

**`map` applies a function to every element of a sequence and returns an iterator:**

```python
simpsons = ['homer', 'marge', 'bart']
map(len, simpsons)
list(map(len, simpsons))
```

```python
# equivalent list comprehension
[len(word) for word in simpsons]
```

```python
map(lambda word: word[-1], simpsons)
list(map(lambda word: word[-1], simpsons))
```

```python
# equivalent list comprehension
[word[-1] for word in simpsons]
```

**`filter` returns a iterator containing the elements from a sequence for which a condition is `True`:**

```python
nums = range(5)
filter(lambda x: x % 2 == 0, nums)
```

```python
# equivalent list comprehension
[num for num in nums if num % 2 == 0]
```
